```python
n1 = 10
if n1 % 2:
    print(n1,"odd")
else:
    print(n1,"even")
```

```python
n2 = 10
if n2 % 2:
    print("%d odd" % (n2))
else:
    print("%d even" %(n2))
```

```python
n2 = 11
if n2 % 2:
    print("{} odd".format(n2))
else:
    print("{} even".format(n2))
```

```python
n3 = 10
if n3 % 2:
    print(f"{n3} odd")
else:
    print(f"{n3} even")
```
