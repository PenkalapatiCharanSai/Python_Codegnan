### Output formatting

### 1. Using old style formatting using comma

```python
age = int(input())
name = input()
print("My name is", name ,"and I am", age ,"years old")
```

```python
age = int(input())
name = input()
print("My name is "+ name +" and I am "+ age +" years old")
```

```python
age = int(input())
name = input()
print("My name is "+ name +" and I am "+ str(age) +" years old")
```

### 2. Using modulo operator

```python
age = int(input())
name = input()
print("My name is %s and I am %d years old" % (name, age))
```

```python
age = int(input())
name = input()
print("My name is %s and I am %s years old" % (name, age))
```

```python
age = int(input())
name = input()
print("My name is %d and I am %s years old" % (name, age))
```

### 3. By Using Dot format method

```python
age = int(input())
name = input()
print("My name is {} and I am {} years old".format(name, age))
```

```python
age = int(input())
name = input()
print("My name is {} and I am {} years old".format(age, name))
```

### 4. Using f-format method

```python
age = int(input())
name = input()
print(f"My name is {name} and I am {age} years old")
```

```python
age = float(input())
name = input()
print(f"My name is {name} and I am {age:.2f} years old")
```
