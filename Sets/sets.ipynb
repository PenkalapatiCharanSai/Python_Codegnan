#### Set Methods
- add(val) : It adds the value to set of the values is already not exists in set
- update(new_set) : It adds all new set elements
- remove(val) : It removes the specified value from the set if it is present in set otherwise``` KeyError```
- discard(val) : It removes the value if it present in set, otherwise nothing
- union(new_set) : It returns the all unique from sets.
- intersection(new_set) : It return the only common elements from all sets
- difference(new_set) : It returns the set elements those are not presnet in new_set
- symmetric_differnce(new_set) : It returns the common elements from the sets

````python
s = {1,2,3,4,5}
p = {4,5,6,2,3}
print(s,p)
````

````python
s.add(10)
print(s)
````

````python
s.update({1,2,100,200})
print(s)
````

````python
# remove(val), discard(val)
s = {1,2,3,4,5,10}
s.remove(2)
print(s)
s.remove(2) # It raises key error
````

````python
s.discard(10)
print(s)
s.discard(10)
print(s) # It won't raise any error
````
