```python
## Scope of Variable
#1. Global Scope
#2. Local Scope
#3. Non-local Scope
```

```python
x=10
def outer(a):
    c=a+10
    def inner(b):
        k=b+10
        print("This is from inner function:", k)
    inner(c)
    print("This is from Outer Function:", c)
outer(x)
print(x)
```

```python
#Global variables using in local and non-local scope:-
x=10
def outer(a):
    c=a+x
    def inner(b):
        k=c+x
        print("This is from inner function:", k)
    inner(c)
    print("This is from Outer Function:", c)
outer(x)
print(x)
```

```python
#Global variable trying to update in local scope by using global variable

x=10
def outer(a):
    global x
    x=a+x
    c=x+2
    def inner(b):
        nonlocal c
        c=c+b
        print("This is from inner function:", c)
    inner(c)
    print("This is from Outer Function:", c)
outer(x)
print(x)
```

## Pass By Value & Pass By Reference

### Pass by Value:- In pass by value, the argument objects share the object copy value to function definition parameter.
### All immutable datatypes (i.e string, tuple, int) behave like pass by value
### After function execution, the original value not effects

### Pass by Reference:- In pass by reference, the argument objects shares the object (argument variable) reference to function definition parameter. So, after the function execution, original values will effects.
### All mutable datatypes(i.e list, set, dictionary) behave like pass by reference

```python
#Passs By Value:-
a='Praharsha'
def replace(string, old, new="@"):
    res=string.replace(old,new)
    return res
print(replace(a,'a', 'v'))
print(a)    
    
```

```python
#Pass By Reference
lst = [1,2,3,4]
def add_ele(arr):
    arr.append(100)
    return arr
print(add_ele(lst))
print(lst)
```

## Built-in Functions:-
### 1. Data Type Functions: int(),float(), str(), complex(), bool(), list(), tuple(), set(), dict()
### 2. abs(): It return absolute value (+ve value)
### 3. ord(): It returns ASCII value of a character
### 4. chr(): It returns character for specied ASCII valule
### 5. len(): It returns total number of elements in sequence
### 6. min(): It returns minimum value from sequence
### 7. max(): It returns maximum value from the sequence
### 8. sum(): It adds all the elements when the sequence contains numeric elements
### 9. sorted(): This function returns the sequence either 
### 10. enumerate(): It returns the index values along with sequence elements
### 11. round(): It rounds the floating values
### 12. filter(fun, seq): It filter the sequence values 
### 13. map(fun, seq): It applies functionality on each element in sequence
### 14. zip(): It returns multiple sequences as a single sequence

```python
print(round(10.2357, 2))
print(f"{10.2357:.2f}")
```

```python
def even(x):
    if x%2==0:
        return x
lst = [1,2,3,4,5,6]
print(list(filter(even, lst)))
```

```python
def add(x,y):
    return x+y
print(list(map(add, [1,2,3], [1,2,3])))
print(list(map(add, [1,2,3,4,5,6], [1,2,3])))
```
