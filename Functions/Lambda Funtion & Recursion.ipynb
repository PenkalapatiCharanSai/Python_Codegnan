### Lambda Funtion:- It does not have function names and we can say it as anonymous function. This function are implemented by lambda keyword.

```python
#Syntax:-
## lambda(keyword)  arguments(variable name) : Expression
```

```python
#Anonymous function
#Find the square of a number

square = lambda x: x*x
print(square(2))
print(type(square))
```

```python
#Single line if-else statement

#Syntax:- True-Exp if condition else False-Exp
```

```python
#Find weather a number is Even or Odd
res = lambda x: "Even" if x%2==0 else "Odd" 
print(res(5))
```

```python
#return square root of a number, if number is even otherwise square of a number
res = lambda x: x**0.5 if x%2==0 else x*x
print(int(res(5)))
print(int(res(6)))
```

```python
#check weather a number is perfect square or not
res = lambda x: "Perfect Square" if x**0.5 == int(x**0.5) else "Not Perfect Square"
print(res(10))
print(res(9))
print(res(55))
```

```python
#Filter the even numbers from the list using lambda and filter function
lst = [1,2,3,4,5,6]
print(list(filter(lambda x: x%2==0, lst)))
```

```python
#Find the square of each number in list using map, lambda functions
lst = [1,2,3,4,5,6]
print(list(map(lambda x: x*x, lst)))
```

### Recursion:- Function calls itself either directly or inderectly is known as recursion
### Types of Recursion:-
### 1. Direct Recursion
### 2. Indirect Recursion

### Direct Recursion:- Function calls itself
#### Here, mainly it have two blocks:-
#### 1. Base Case
#### 2. Recursive Call Case


### Indirect Recursion:- Function A calls fuction B and Function B calls function A. We can call it as cycle



```python
#Direct Recursion
def fact(n):
    if n==0 or n==1:
        return 1
    return n*fact(n-1)
n = int(input())
print(fact(n))
```

```python
#print n to 1 numbers using recursion
def print_to_1(n):
    if n==1:
        print(n)
        return
    print(n, end = " ")
    print_to_1(n-1)
print_to_1(5)
```

```python
#print 1 to n numbers using recursion
def print_to_n(n):
    if n==1:
        print(n, end=" ")
        return
    print_to_n(n-1)
    print(n,end=" ")
n = int(input())
print_to_n(n)
```

```python
#print 1 to n numbers in list using recursion
def list_to_n(n):
    if n == 0:
        return []
    result = list_to_n(n - 1)
    result.append(n)
    return result

n = int(input("Enter n: "))
print(list_to_n(n))
```

```python
#print n to 1 numbers in list using recursion
def list_n_to_1(n):
    if n == 0:
        return []
    return [n] + list_n_to_1(n - 1)

n = int(input("Enter n: "))
print(list_n_to_1(n))
```
