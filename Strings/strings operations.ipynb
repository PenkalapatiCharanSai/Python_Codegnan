####String methods
 - case conversion methods
    - upper() : It returns upper case converted string
    - lower() : It covert entire string into lower state
    - title() : 
    - swapcase()
    - capitalize()

```python
s = "The sky is blue and Ocean is BLUE"
print(s.upper())
print(s.lower())
print(s.title())
print(s.swapcase())
print(s.capitalize())
```

```python
s = "25"
print(s.upper())
print(s.lower())
print(s.title())
print(s.swapcase())
print(s.capitalize())
```

```python
# substring index finding
# find (Substring): It returns the first accurance index position of substring, substring, otherwise -1
# index(substring) 
s = "Codegnan"
print(s.find('n'))
print(s.find('gn'))
print(s.find('code'))
print(s.find('N'))
print(s.index("N"))
```

-***white space or pattern removing from string***
- strip(chars) : Return a copy of the string with leading and trailing whitespace removed.
- lstrip() : Return a copy of the string with leading whitespace removed.
- rstrip() : Retuen a copy of the string with trailing whitespace removed.

```python
s = "      codegnan        "
print(s)
print(s.strip())
print(s.lstrip())
print(s.rstrip())
```

```python
s = "   @@@@#@@@ Code gnan @@   "
print(s.strip())
print(s.lstrip())
print(s.rstrip())
```

```python
s = "@@@@ #@ Code gnan @@"
print(s)
print(s.strip('@'))
print(s.lstrip('@'))
print(s.rstrip('@'))
```

### String Manipulation Methods
- replace(old, new) : it replaces old string with new string
- split() : it splits string into of substrings
- "sep".join(list) : it combines the all list of words with seperator


```python
s = "The sky is blue and Ocean is BLUE"
print(s.replace('is','ISs'))
print("ABCBCBC".replace('CBC', '@'))
```

```python
print(s.split())
```

```python
print(s.split('blue'))
```

```python
p = "Charan"
c = "Sai"
print(p.join(c))
```

```python
words = ['The', 'sky', 'is', 'blue', 'and', 'Ocean', 'is', 'BLUE']
print("".join(words))
print(" ".join(words))
print("_".join(words))
```

#### String checking methods
 - This all methods return boolean values as result
    - islower() : it returns true if entire string is lower
    - isupper() : it returns true if entire string is uppercase
    - istitle() : it returns true if entire string is title manner
    - isalpha() : it returns true if entire string is contains only alphabets
    - isallnum() : it returns true if entire string is contains  alphabets and digits
    - isdigit() : it returns true is entire string contains only digits
    - startswith(sub_string) : it returns true if string startswith sub)string
    - endswith(sub_string) : it returns true if string endswith_sub_string
