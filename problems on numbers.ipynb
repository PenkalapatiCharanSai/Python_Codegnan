```python
# Find length of number
def length_number(n):
    count = 0
    while n > 0:
        digit = n % 10
        n  //= 10
        count += 1
    return count
n = int(input())
print(length_number(n))
```

```python
# Find the sum of digits
def sum_of_digits(n):
    sum = 0
    while n > 0:
        digit = n % 10
        sum += digit
        n //= 10
    return sum
n = int(input())
print(sum_of_digits(n))
```

```python
#### Find reverse of a number
def reverse_number(n):
    rev = 0
    while n > 0:
        digit = n % 10
        rev = rev * 10 + digit 
        n //= 10
    return rev
n = int(input())
print(reverse_number(n))
```

```python
# Armstrong number

def is_armstrong(n):
    temp = n

    # Step 1: count digits
    count = 0
    t = n
    while t > 0:
        count += 1
        t //= 10

    # Step 2: calculate sum of powers
    total = 0
    t = n
    while t > 0:
        digit = t % 10
        total += digit ** count
        t //= 10

    return total == temp


n = int(input())
if is_armstrong(n):
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

```python
# sum of digits using recursion
def sum_of_digits(n,res = 0):
    if n == 0:
        return res
    return sum_of_digits(n // 10,res + n % 10)
n = int(input())
print(sum_of_digits(n))
```
