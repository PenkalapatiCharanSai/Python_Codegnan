```python
# Find length of number
def length_number(n):
    count = 0
    while n > 0:
        digit = n % 10
        n  //= 10
        count += 1
    return count
n = int(input())
print(length_number(n))
```

```python
# Find the sum of digits
def sum_of_digits(n):
    sum = 0
    while n > 0:
        digit = n % 10
        sum += digit
        n //= 10
    return sum
n = int(input())
print(sum_of_digits(n))
```

```python
#### Find reverse of a number
def reverse_number(n):
    rev = 0
    while n > 0:
        digit = n % 10
        rev = rev * 10 + digit 
    return rev
n = int(input())
print(reverse_number(n))
```
