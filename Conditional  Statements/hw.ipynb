```python
#write a program to add two numbers and print the result
a = int(input())
b = int(input())
c = a + b
print("sum of a and b is",c)
```

```python
#Take two integers as input and print their quotient and remainder.
a = int(input())
b = int(input())
q = a // b
r = a % b
print("quotient of",a,"and",b,"is",q)
print("remainder of",a,"and",b,"is",r)
```

```python
#Given a number, check if it is even or odd using the modulus operator.
a = int(input())
if a % 2 == 0:
    print("even")
else:
    print("odd")
```

```python
#Write a program to swap two variables using arithmetic operators (without a third variable).
a = int(input("Enter a value:"))
b = int(input("Enter b value:"))
a,b = b,a
print("After swaping")
print("a value is",a,"and"," b value is", b)
```

```python
#Take three numbers and print the largest using comparison operators.
a = int(input("Enter a value"))
b = int(input("Enter b value"))
c = int(input("Enter c value"))
if a > b and a > c:
    print(a,"is greater than",b,"and",c)
elif b > a and b > c:
    print(b,"is greater than",a,"and",c)
else:
    print(c,"is greater than",a,"and",b)
```

```python
#Write a program to calculate the square of a number using the exponent operator （**)
a = int(input("Enter a value"))
b = int(input("Enter b value"))
c = a ** b
print("exponent of",a,"and",b,"is",c)
```

```python
#Given two numbers, check if the first is divisible by the second.
num1 = int(input("Enter a number"))
num2 = int(input("Enter b number"))
if num2 == 0:
    print("Cannot divide by zero")
elif num1 % num2 == 0:
    print(num1,"is divisible by",num2)
else:
    print(num1,"is not divisible by",num2)
```

```python
#Write a progran to calculate the area of a rectangle using multiplication operator.
length = float(input("Enter length of rectangle"))
width = float(input("Enter width of rectangle"))
area = length * width
print("area of rectangle is",area)
```

```python
#Take a nunber and print whether it is positive, negative, or zero.
num = int(input("Enter a number"))
if num > 0:
    print(num,"is positive")
elif num < 0:
    print(num,"is negative")
else:
    print(num,"is zero")
```

```python
#Write a program to check if two numbers are equal using the equality operator (==).
a = int(input("Enter a value"))
b = int(input("Enter b value"))
if a == b:
    print(a,"and",b,"equal")
else:
    print(a,"and",b,"not equal")
```
