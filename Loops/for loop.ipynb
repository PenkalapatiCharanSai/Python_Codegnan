## for loop

```python
n = int(input())
for num in range(1,n+1):
    print(f"{num} - {num ** 2}")
print("done")
```

```python
n = int(input())
for num in range(1,11):
    print(f"{n} x {num} = {n * num}")
```

```python
n1 = int(input())
n2 = int(input())
even = []
odd = []
for num in range(n1,n2+1):
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)
print(f"Even numbers {even}")
print(f"Odd numbers {odd}")
```

```python
n1 = int(input())
n2 = int(input())
even = []
odd = []
for num in range(n1,n2+1):
    if num % 2:
        odd.append(num)
    else:
        even.append(num)
print(even)
print(odd)
```

```python
n1,n2 = map(int, input().split())
even = []
odd = []
for num in range(n1,n2+1):
    if num % 2:
        odd.append(num)
    else:
        even.append(num)
print(f"even {even}")
print(f"odd {odd}")
```

```python
# Find the factors of a number
n = int(input())
factors = []
for num in range(1,n+1):
    if n % num == 0:
        factors.append(num)
print(f"factors of {n} is {factors}")
```

```python
# Find the factors of a number
n = int(input())
factors = []
for num in range(1,n//2+1):
    if n % num == 0:
        factors.append(num)
factors.append(n)
print(f"factors of {n} is {factors}")
```

```python
# Sum of n natural numbers
n = int(input())
result = 0
for num in range(1,n+1):
    result += num
    #num += 1
print(f"Sum of {n} natural numbers is {result}")
```

```python
# Sum of n natural numbers
n = int(input())
res = 0
if n >= 1 and n <= 9:
    for num in range(1,n+1):
        print(num, end = "")
        res += num
else:
    print("Invalid")
print(f"\n{res}")
```

```python
#Check weather target element is exists in list of elements or not
#indexed based
target = int(input())
arr = list(map(int, input().split()))
for num in range(len(arr)):
    if arr[num] == target:
        print(f"{target} exists in {arr}")
        break
else:
    print(f"{target} not exists in {arr}")
```

```python
#element based 
target = int(input())
arr = list(map(int, input().split()))
for num in arr:
    if num == target:
        print(f"{target} exists in {arr}")
        break
else:
    print(f"{target} not exists in {arr}")
```

```python
# Factorial
n = int(input())
res = 1
if n == 0:
    print(res)
else:
    for i in range(1, n + 1):
        res *= i
    print(res)
```

```python
# Factorial
n = int(input())
res = 1
if n == 0 or n == 1:
    print(res)
else:
    for i in range(2, n + 1):
        res *= i
    print(res)
```
