```python
n = int(input())
num = 1
while num < n + 1:
    print(num, end = " ")
    num += 1
```

```python
n1, n2 = map(int, input().split())
even = []
odd = []
while n1 <= n2:
    if n1 % 2 == 0:
        even.append(n1)
        n1 += 1
    else:
        odd.append(n1)
        n1 += 1
print(even)
print(odd)
```

```python
n1, n2 = map(int, input().split())
even = []
odd = []
i = n1 # Initialization
while i <= n2:
    if i % 2 == 0:
        even.append(i)
        i += 1
    else:
        odd.append(i)
        i += 1
print(even)
print(odd)
```
