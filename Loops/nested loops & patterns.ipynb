### Nested Loops and patterns
### A loop contains another loop


```python
for i in range(1,4):
    for j in range(1,4):
        print(i,j)
```

```python
n = 5
for i in range(n):
    for j in range(n):
        print((i, j), end = " ")
```

```python
n = 5
for i in range(n):
    for j in range(n):
        print((i, j), end = " ")
    print()
```

```python
n = 5
for i in range(n):
    for j in range(n):
        print(("*"), end = " ")
    print()
```
