```python
import math
print(math.ceil(13.07)) #it returns the upperbound value
print(math.floor(13.07)) #it returns the lowerbound value for a given number)
print(math.factorial(5)) #it returns factorial of a number
print(math.pow(10,3)) #pow(a,b) is, it retutrns the a power of b
print(math.gcd(10, 5, 25)) #it returns the greatest common factors
print(math.pi) #It returns the pi value
print(math.isqrt(10)) #it returns the square root of a number in integer frmat i.e without decimals
```
