```python
# inheritence paerent class constructor in child class by using super()
# super(): This function is used to access the present class properties is current child class 
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return {"Name": self.name, "Age": self.age}
class Student(Person):
    def __init__(self, stdid, name, age):
        self.stdid = stdid
        super().__init__(name, age)
        
p = Person("Charan",20)
s = Student(101,"sai",21)
print(s.details())
print(s.name)
print(s.age)
```

```python
# multilevel inheritence 
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return {"Name": self.name, "Age": self.age}
class Student(Person):
    def __init__(self, id, name, age):
        self.id = id
        super().__init__(name,age)
    def show_details(self):
        return {"id" : self.id,"name" : self.name, "age":self.age}
class Employee(Student):
    def __init__(self, id,name,age,salary):
        self.salary = salary
        Student.__init__(self,id,name,age)
    def display_details(self):
        return {"id": self.id, "name":self.name,"age":self.age,"salary":self.salary}
p = Person("charan",20)
s = Student(459, "sai",21)
e = Employee(102,"charan sai",25,50000)
print(e.details())
print(e.show_details())
print(e.display_details())
```

```python
# multilevel inheritence 
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return {"Name": self.name, "Age": self.age}

        
class Student(Person):
    def __init__(self, id, name, age):
        self.id = id
        super().__init__(name,age)
    def show_details(self):
        user = super().details()
        user['id'] = self.id
        return user

        
class Employee(Student):
    def __init__(self, id,name,age,salary):
        self.salary = salary
        Student.__init__(self,id,name,age)
    def display_details(self):
        std = super().show_details()
        std['salary'] = self.salary
        return std
p = Person("charan",20)
s = Student(459, "sai",21)
e = Employee(102,"charan sai",25,50000)
print(e.details())
print(e.show_details())
print(e.display_details())
```

```python
# multiple inheritance

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return {"name": self.name, "age": self.age}


class Student:
    def __init__(self, id, name, age):
        self.id = id
        self.name = name
        self.age = age
    def student_details(self):
        return {"id": self.id, "name": self.name, "age": self.age}


class Employee(Person, Student):
    def __init__(self, id, name, age, salary):
        Student.__init__(self, id, name, age)
        self.salary = salary
    def employee_details(self):
        print(super().details()) 
        return {"id": self.id, "name": self.name, "age": self.age, "salary": self.salary}


e = Employee(101, "Charan", 25, 50000)
print(e.employee_details())
print(Employee.__mro__) # method resulation order
```
