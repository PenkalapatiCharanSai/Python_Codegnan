```python
from abc import ABC, abstractmethod

class Mobile(ABC):

    @abstractmethod
    def ram(self):
        pass

    @abstractmethod
    def rom(self):
        pass

    @abstractmethod
    def battery(self):
        pass


class Oppo(Mobile):

    def __init__(self, ram, rom, battery_capacity):
        self.ram_size = ram
        self.rom_size = rom
        self.battery_capacity = battery_capacity
    def ram(self):
        return f"This mobile contains {self.ram_size} GB RAM"
    def rom(self):
        return f"This mobile contains {self.rom_size} GB ROM"
    def battery(self):
        return f"This mobile Battery capacity is {self.battery_capacity} mAh"
    def sim_slot(self):
        return "This mobile have (1 sim and 1 memory slot) or 2 sim slots"
# m = Mobile()
reno10 = Oppo(8, 256, 5000)
remo = Oppo(6, 128, 6000)
print(reno10.ram())
print(remo.ram())
```

```python
# 2D and 3D shapes in Abstraction
import math
from abc import ABC, abstractmethod

class Shape2D(ABC):
    @abstractmethod
    def area(self):
        pass
        
class Shape3D(ABC):
    @abstractmethod
    def volume(self):
        pass
        
class Square(Shape2D):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side * self.side
        
        
class Rectangle(Shape2D):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
        
        
class Cube(Shape3D, Shape2D):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side * self.side
    def volume(self):
        return self.side * self.side * self.side
        
class Cone(Shape3D, Shape2D):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height
    def area(self):
        return 2 * math.pi * self.radius * (self.radius + self.height)
    def volume(self):
        return 1/3 * math.pi * self.radius * self.radius * self.height

        
class Sphere(Shape3D, Shape2D):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        sphere_area = 4 * math.pi * self.radius * self.radius
        return sphere_area
    def volume(self):
        sphere_volume = 4/3 * math.pi * self.radius * self.radius * self.radius
        return sphere_volume

        
class Pyramid(Shape3D, Shape2D):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def area(self):
        return self.base * self.height
    def volume(self):
        return 1/3 * self.base * self.base * self.height

        
square1 = Square(5)
square2 = Rectangle(5, 10)
cube1 = Cube(10)
cube2 = Cone(20, 30)
sphere3 = Sphere(10)
print(square1.area())
print(square2.area())
print(cube1.area())
print(cube2.area())
print(sphere3.area())
print(cube1.volume())
print(cube2.volume())
print(sphere3.volume())
```
