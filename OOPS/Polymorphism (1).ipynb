### Method Overriding

```python
# Method overriding

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return {"Name": self.name, "Age": self.age}
class Student(Person):
    def __init__(self, name, age, id):
        super().__init__(name,age)
        self.id = id
    def details(self):
        return {"Name": self.name, "Age": self.age, "ID":self.id}
s = Student("Charan",20,459)
print(s.details())
```

```python
# execute the both methods with single call
# Method overriding

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return {"Name": self.name, "Age": self.age}
class Student(Person):
    def __init__(self, name, age, id):
        super().__init__(name,age)
        self.id = id
    def details(self):
        user = super().details()
        user.update({"ID":self.id})
        return user
s = Student("Charan",20,459)
print(s.details())
```

### Method Overloading

```python
class Math:
    def add(self, x, y):
        return x + y
    def add(self, x, y, z):
        return x + y + z
    def add(self, w, x, y, z):
        return w + x + y + z
m = Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

### Python doesn't support Method overloading but, we can achive the same functionality through the default arguments or variable length of positional of VLKW

```python
# Method overloading
class Math:
    def add(self, w, x, y = 0, z = 0):
        return w + x + y + z
m = Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
# Magic method and Operator Overloading
class Add:
    def __init__(self, x):
        self.x = x
a1 = Add(10)
# print(a1)
a2 = Add(10)
print(a1 == a2)
print(a1.x + a2.x)
# print(a1 + a2)
print(dir(a1))
```

```python
# if class contains __eq__magic method
class Add:
    def __init__(self, x):
        self.x = x
    def __eq__(self, other):
        return self.x == other.x
    def __add__(self, other):
        return self.x + other.x
    def __sub__(self, other):
        return self.x - other.x
    def __mul__(self, other):
        return self.x * other.x
        
a1 = Add(10)
# print(a1)
a2 = Add(10)
print(a1 == a2)
print(a1 + a2)
print(a1 - a2)
print(a1 * a2)
```

```python
# if class contains __eq__magic method
class Add:
    def __init__(self, x):
        self.x = x
    def __eq__(self, other):
        return self.x == other.x
    def __add__(self, other):
        return self.x + other.x 
    def __sub__(self, other):
        return self.x - other.x
    def __mul__(self, other):
        return self.x * other.x
        
a1 = Add(10)
a2 = Add(10)
res = a1 + a2
a3 = Add(30)
print(res + a3.x)
```

```python
# if class contains __eq__magic method
class Add:
    def __init__(self, x):
        self.x = x
    def __eq__(self, other):
        return self.x == other.x
    def __add__(self, other):
        return Add(self.x + other.x) 
    def __sub__(self, other):
        return self.x - other.x
    def __mul__(self, other):
        return self.x * other.x
    def __str__(self):
        return str(self.x)
        
a1 = Add(10)
a2 = Add(10)
res = a1 + a2
a3 = Add(30)
print(res + a3)
```

```python
# if class contains __eq__magic method
class Add:
    def __init__(self, x):
        self.x = x
    def __str__(self):
        return f"{self.x}"
    def __add__(self, other):
        return Add(self.x + other.x) 
        
a1 = Add(10)
a2 = Add(10)
res = a1 + a2
print(res)
a3 = Add(30)
print(res + a3+Add(100))
```
