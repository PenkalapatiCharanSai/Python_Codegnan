```python
class Student:
    school = "Codegnan" #class attribute
    def __init__(self,name,age,school):
        self.name = name # instance attribute
        self.age = age # instance attribute
        self.school = school

    def details(self):
        return f"Myname is {self.name} and i am {self.age} years old"
    def percentage(self,marks:list):
        return sum(marks)/3
#object creation
s1 = Student("Ravi", 20, "ABC School")
print(s1)
print(s1.details())
print(s1.percentage(list([45,65,85])))
```

```python
class Student:
    school = "Codegnan" #class attribute
    def __init__(self,name,age):
        self.name = name # instance attribute
        self.age = age # instance attribute
    

    def details(self): # instance method
        return f"Myname is {self.name} and i am {self.age} years old {Student.school}"

    @classmethod
    def update_school_name(cls, new_name): # class method
        cls.school = new_name

    @staticmethod
    def percentage(marks:list): # static method
        return sum(marks)/3

    
#object creation
s1 = Student("Ravi", 20)
s2 = Student("suresh", 22)
print(s1)
print(s1.details())
print(s1.update_school_name("Hello new school"))
print(s1.school)
print(s2.school)
Student.school = "XYZ School"
print(s1.school)
print(s2.school)
```
