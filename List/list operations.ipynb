# list methods
- append(ele): It add the element at the end of the list
- extend(list_ele) : It adds all list elements at end of list
- insert(pos,val) : It inserts the value at particulat position
## removing elements from list
- remove(val) : It removes the first occurance value if value is present in list, otherwise ValueError
- pop() : It remove and returns the last element in the list
- pop(pos) : It removes and returns specified position element from the list.
## index finding for an element
- index(val) : It returns the first occurance value index position
## sorting list
- sort() : It sorts the original list into asscending order
- sort(reverse = True) : It sorts athe original list into descending order

```python
# removing elements from list
lst = [10,2,3,4,2,50]
lst.remove(2)
print(lst)
#lst.remove(100) # it raise value error because, value not in the list
val = lst.pop()
print(val,lst)
lst.clear # it removes all elements from the list
print(lst)

```

```python
l1 = [1,2,3]
l2 = [1,2,3]
print(l1 == l2)
print(l1 is l1)
```

```python
# shallow copy
l1 = [1,2,3]
l2 = l1
print(l1 == l2)
print(l1 is l1)
l1.append(10)
print(l1)
```

```python
# deep copy
l1 = [1,2,3]
l2 = l1[:]
print(l1 == l2)
print(l1 is l1)
l1.append(10)

print(l2)
```

```python
# built-in functions
# len(), min(), max(), sum()

lst = [10, 20, 30, 40, 'hi']

# length works fine
print("Length:", len(lst))

# min, max, sum will give error because of string 'hi'
# So we filter only integers

nums = [x for x in lst if isinstance(x, int)]

print("Min:", min(nums))
print("Max:", max(nums))
print("Sum:", sum(nums))
```

```python
lst = [10, 20, 30]

lst.append(40)
lst.extend([50, 60])
lst.insert(1, 15)

print("After adding:", lst)

lst.remove(20)
print("After remove:", lst)

print("Popped:", lst.pop())
print("Popped at index 1:", lst.pop(1))

print("Index of 30:", lst.index(30))

lst.sort()
print("Ascending:", lst)

lst.sort(reverse=True)
print("Descending:", lst)
```
