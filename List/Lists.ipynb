```python
# list is ordered collection
lst = [1,2,3,1,2,2.5,'HI']
print(lst)
```

```python
# list is supports indexing adn slicing
# 
print(lst[-2])
print(lst[-2:3:-1])
```

```python
print(lst[3:-2])
```

```python
print(lst[::-1]) # reverse list using slicing
```

```python
# list is mutable 
lst[1] = 100
print(lst)
```

```python
# empty list representation
l1 = []
l2 = list()
print(l1,l2)
print(type(l1),type(l2))
```
